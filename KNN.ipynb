```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
def normalization(x):
 mu=np.mean(x,axis=0)
 stnd=np.std(x,axis=0)
 stnd=np.where(stnd==0,1e-10,stnd)
 return mu,stnd
```

```python
def prediction(x2,x,y):
 predictions=(np.zeros(m,int)).T
 distances=np.zeros(xl_train.shape[0],)
 for i in range(m):
  distances=np.sqrt(((x2[i,:]-x)**2).sum(axis=1))
  k_nearest_indices = np.argsort(distances)[:k]
  k_nearest_labels = y[k_nearest_indices]
  predictions[i]=np.argmax(np.bincount(k_nearest_labels))
 return(predictions)
```

```python
dfl=np.array(pd.read_csv("Classification_train.csv"))
np.random.shuffle(dfl)
xl=dfl[:,1:]
mu,stnd=normalization(xl)
xl= (xl-mu)/stnd
m,n = xl.shape
a=int(70/100*m)
xl_train=xl[:a,:]
```

```python
xl_validate=xl[a:,:]
m,n=xl_validate.shape
yl=dfl[:,0]
yl_train=yl[:a]
yl_validate=yl[a:]
```

```python
k=6
predictions=prediction(xl_validate,xl_train,yl_train)
print(f"Accuracy: {np.mean(yl_validate==predictions)}")
df1=np.array(pd.read_csv("Classification_test.csv"))
x1=df1[:,1:]
x1=(x1-mu)/stnd
m,n = x1.shape
Id=df1[:,0]
predictions=prediction(x1,xl_train,yl_train)
for i in range(len(Id)):
 print(f"ID:{Id[i]}  label:{predictions[i]}" )
```
